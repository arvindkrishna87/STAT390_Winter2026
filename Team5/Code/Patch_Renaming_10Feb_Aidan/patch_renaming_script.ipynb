```python
import os
import re
```

```python
folder_path = "/projects/e32998/patches_Team5"

files = os.listdir(folder_path)

old_case_number = 25      # case to change
new_case_number = 119      # new case number
desired_match = 4        # ONLY change files with this match
desired_stain = "h&e"   # choose: "h&e", "melan", "sox10"
```

```python
pattern = rf"(case_){old_case_number}(?=_)(?=.*unmatched_{desired_match}(?=_|$))(?=.*{desired_stain})"

for filename in files:
    new_name = re.sub(
        pattern,
        rf"\g<1>{new_case_number}",
        filename
    )

    if new_name != filename:
        print(f"{filename} -> {new_name}")
        os.rename(
            os.path.join(folder_path, filename),
            os.path.join(folder_path, new_name)
        )
```

```python
finder_pattern = rf"(case_){old_case_number}(?=_)(?=.*unmatched_{desired_match}(?=_|$))(?=.*{desired_stain})"

filtered_list = [name for name in files if re.search(finder_pattern, name)]

filtered_list
len(filtered_list)
```
